### Load packages

```python
from pathlib import Path
import scanpy as sc
import liana as li
import pandas as pd
```

### Convert lrdata to muData

```python
lrdata_path  = Path("../../data/MERFISH_mouse_brain/")
adata = sc.read_h5ad(lrdata_path / "lrdata_C57BL6J_2.039.h5ad")
```

```python
mdata = li.utils.anndata_to_mdata(
    adata = adata,
    groupby = "cell_type",
    min_cells=10,
    by_var = True,
    verbose = True)
```
